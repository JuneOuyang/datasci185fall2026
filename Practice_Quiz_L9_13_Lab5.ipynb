### Question (A) — Dictionaries and a running maximum
*Material from Lecture 9*

The dictionary `reading_club` has one key for each student. Each value is a dictionary containing that student's year in college and the number of pages they read.

Using a **`for` loop and a running maximum**, find the student who read the most pages.

Before the loop, the variables `most_pages = 0` and `top_reader = "nobody"` are provided. Whenever you find a larger page count, update **both** variables. You may assume there is one unique highest page count and all page counts are positive.

Do not use `max()`, `np.max()`, or sorting to find the winner. Complete the loop so that the final print statement reports the correct name and page count.

````python
reading_club = {
    "Maya":   {"college_year": 2, "pages": 185},
    "Noah":   {"college_year": 1, "pages": 240},
    "Lina":   {"college_year": 3, "pages": 165},
    "Ethan":  {"college_year": 2, "pages": 310},
    "Sofia":  {"college_year": 4, "pages": 275},
    "Daniel": {"college_year": 1, "pages": 205}
}
most_pages = 0
top_reader = "nobody"

# your for loop here
for reader in reading_club:
    pages_read = reading_club[reader]["pages"]
    if pages_read > most_pages:
        most_pages = pages_read
        top_reader = reader

print(f"The top reader is {top_reader}, with {most_pages} pages.")
````

### Question (B) — Averages from nested dictionaries
*Material from Lecture 9*

The set-up cell defines weekly visitor counts for five museums. It randomly selects **2 to 4 museums** and stores their data in `museums_to_average`.

Using **`for` loops**, compute the average number of visitors on each day across the selected museums. Print one line per day in the form:

```text
Average visitors on Monday: 140.00
```

Display each average with **two decimal places**. Your code must work for any selection and must use the actual number of selected museums.

For example, if only Art and History are selected, the Monday average is `(120 + 160) / 2 = 140.00`.

Run the set-up cell several times to test different selections. Do not use a DataFrame to solve this question.

````python
# Set-up code — do not change
import random

museum_visitors = {
    "Art":     {"Monday": 120, "Tuesday": 135, "Wednesday": 150, "Thursday": 145},
    "History": {"Monday": 160, "Tuesday": 155, "Wednesday": 180, "Thursday": 170},
    "Science": {"Monday": 210, "Tuesday": 195, "Wednesday": 225, "Thursday": 240},
    "Nature":  {"Monday": 140, "Tuesday": 150, "Wednesday": 165, "Thursday": 155},
    "Design":  {"Monday": 100, "Tuesday": 115, "Wednesday": 125, "Thursday": 130}
}
days = ["Monday", "Tuesday", "Wednesday", "Thursday"]
num_museums = random.randint(2, 4)
selected_museums = random.sample(list(museum_visitors.keys()), k=num_museums)
museums_to_average = {}
for museum in selected_museums:
    museums_to_average[museum] = museum_visitors[museum]
print("Selected museums:", selected_museums)
# End of set-up code
````

````python
for day in days:
    total_visitors = 0

    for museum in museums_to_average:
        total_visitors += museums_to_average[museum][day]

    average_visitors = total_visitors / len(museums_to_average)

    print(f"Average visitors on {day}: {average_visitors:.2f}")
````

### Question (C) — Turn a while loop into a function
*Material from Lecture 10; assertions from Lecture 13*

The code below calculates how many months it takes to save at least **$500**, starting with **$80**. A fixed amount is deposited at the end of each month. There is no interest.

1. Turn the calculation into a function `months_to_goal(monthly_deposit)` that **returns** the required number of months.
2. At the beginning of your function, add an **`assert`** that the monthly deposit is greater than zero, with a useful error message. You may assume the argument is numeric.
3. Use your function to print the number of months needed for monthly deposits of **$20, $35, and $60**.

For a deposit of $35 per month, the function should return **12**. The balance and month counter must start fresh on every call. Do not simply print the result inside the function instead of returning it.

````python
# Given example — run this to see the original calculation
balance = 80
monthly_deposit = 35
months = 0

while balance < 500:
    balance = balance + monthly_deposit
    months = months + 1

print(months)
````

````python
# Define your function here
def months_to_goal(monthly_deposit):
    balance = 80
    months = 0
    while balance < 500:
        balance = balance + monthly_deposit
        months = months + 1
    return months


# Use the function to print the results here
for deposit in [20, 35, 60]:
    print(f"With a monthly deposit of ${deposit}, it will take {months_to_goal(deposit)} months to reach the goal.")
````

Write `supply_total(order, supply_prices)` that returns the cost of all valid items. Process items in their dictionary order.

- If the store does not sell an item, print `Sorry, we don't sell {item}.` and skip it.
- If a quantity cannot be converted with `int()`, print `Sorry, {quantity_string} isn't a valid quantity.` and skip it.
- Detect both problems using **`try`/`except`**, with the specific exception types. Do not use `if` statements to detect these problems.
- The function must continue processing later items and **return** the total after the loop.

All quantities that can be converted to integers in this question are positive. If an item has both problems, either message is acceptable.

Test your function with the supplied order. It should report the problems with **eraser** and **many**, and return **9.5**. Also call it on an empty order; that should return **0.0**.

````python
# Set-up code — do not change
supply_prices = {"pen": 1.50, "notebook": 4.00, "folder": 2.00, "ruler": 1.00}
stationery_order = {
    "pen": "3",
    "eraser": "2",
    "folder": "many",
    "notebook": "1",
    "ruler": "1"
}

# Define your function here
def supply_total(order,supply_prices):
    total = 0.0
    for item, quantity in order.items():
    
        try:
            total += supply_prices[item] * int(quantity)
        except KeyError:
            print(f"Sorry,we don't sell {item}.")
        except ValueError:
            print(f"Sorry, {quantity} is not a valid quantity.")
        
    return total



# Call and test your function here

print(supply_total(stationery_order, supply_prices))

empty_order = {}
print(supply_total(empty_order, supply_prices))
````

````python
def order_total(order, prices):
    total = 0.0
    for item in order:
        quantity_string = order[item]
        try:
            price = prices[item]
            quantity = int(quantity_string)
        except KeyError:
            print(f"Sorry, we don't sell {item}.")
            continue
        except ValueError:
            print(f"Sorry, {quantity_string} isn't a valid quantity.")
            continue
        total += price * quantity
    return total
````

### Question (F) — Keyword-only option and boundary tests
*Material from Lab 5 and Lecture 13*

The supplied function `euclidean_dist` computes the usual straight-line distance between two Cartesian points `(x, y)`.

Write a function `point_distance` that:

- Accepts two points, `point1` and `point2`, each supplied as a tuple. Allow these two arguments to be passed **positionally**.
- Has a third, **keyword-only** parameter, `manhattan`, whose default value is **`False`**.
- When `manhattan=False`, returns the Euclidean distance by calling the supplied function.
- When `manhattan=True`, returns the Manhattan distance:

$$d = |x_1-x_2| + |y_1-y_2|.$$

You may use Python's `abs()` for absolute values. Both options return a number; neither should only print it.

Then write **three assertions** checking:

1. The default distance from `(0, 0)` to `(3, 4)` is 5.
2. With `manhattan=True`, the distance between those points is 7.
3. With `manhattan=True`, the distance from `(2, -1)` to itself is 0.

Add useful assertion messages. Finally, in a comment, explain why the following call should fail, and name the error type:

```python
point_distance((0, 0), (3, 4), True)
```

Do not leave the invalid call active when running your completed notebook.

````python
# Given function — do not change
import numpy as np

def euclidean_dist(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    return np.sqrt((x1 - x2)**2 + (y1 - y2)**2)

# Define point_distance here
def point_distance(point1,point2,*,manhattan=False):
    if manhattan:
        x1, y1 = point1
        x2, y2 = point2
        return abs(x1 - x2) + abs(y1 - y2)
    else:
        return euclidean_dist(point1, point2)
        


# Write your three assertions here
assert point_distance((0,0),(3,4))==5,"It is default (Euclidean) distance"
assert point_distance((0,0),(3,4),manhattan=True)==7,"It is Manhattan distance"
assert point_distance((2,-1),(2,-1),manhattan=True)==0,"It is Manhattan distance"


# Explain the invalid call here:
#point_distance((0,0),(3,4),True) is invalid because the function point_distance is defined to take a keyword argument manhattan. The correct way to call the function with the Manhattan distance option is to use the keyword argument explicitly, like this: point_distance((0,0),(3,4),manhattan=True).
````
