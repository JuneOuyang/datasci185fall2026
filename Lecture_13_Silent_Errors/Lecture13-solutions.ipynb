````python
def one_birthday_trial(room_size):
    room = []
    for k in range(room_size):
        month = random.choice(months)
        day = random.randint(1, 30)
        room.append([month, day])

    for i in range(room_size):
        for j in range(i + 1, room_size):
            if room[i][0] == room[j][0] and room[i][1] == room[j][1]:
                return True

    return False

def estimate_probability(room_size, num_trials):
    num_matches = 0

    for trial in range(num_trials):
        if one_birthday_trial(room_size):
            num_matches = num_matches + 1

    return num_matches / num_trials
````

- Can a candidate with `-999` ever beat the benchmark and be *chosen* in the selection phase?

**Solution**

- No, the selection phase only picks somebody whose skills beats the benchmark, and the benchmark starts at zero.

- "If nobody ever beats the benchmark, you are stuck with the last person." So we can get someone with -999 skills if they happen to be the last person interviewed.

````python
def run_one_trial_clean(N):

    interview_order = list(clean_skills.keys())
    random.shuffle(interview_order)

    best_of_first_N = 0
    for position in range(N):
        skill = clean_skills[interview_order[position]]
        if skill > best_of_first_N:
            best_of_first_N = skill

    chosen_name = interview_order[-1]
    for position in range(N, len(interview_order)):
        skill = clean_skills[interview_order[position]]
        if skill > best_of_first_N:
            chosen_name = interview_order[position]
            break

    return chosen_name, clean_skills[chosen_name]


num_candidates = len(clean_skills)
num_trials = 200
mean_skills_clean = []

for N in range(num_candidates):
    chosen_skills = []
    for trial in range(num_trials):
        chosen_name, chosen_skill = run_one_trial_clean(N)
        chosen_skills.append(chosen_skill)
    mean_skills_clean.append(np.mean(chosen_skills))

# a postcondition on the whole analysis
assert np.min(mean_skills_clean) >= 0, "an average of scores out of 10 cannot be negative"
assert np.max(mean_skills_clean) <= 10, "an average of scores out of 10 cannot exceed 10"

labeled_bar_plot(x_vals = np.arange(num_candidates), heights = mean_skills_clean,
    xlabel = "N", ylabel = "average lab skills of chosen partner",
    title = "Average chosen skill vs. N (cleaned file, 47 candidates)")

print("best average:", round(np.max(mean_skills_clean), 2))
````
