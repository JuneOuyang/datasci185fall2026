# <span style="color:red"> The Lab Partner Problem </span>

<font size="5">

You need to choose a lab partner, and you want the one with the best lab skills (a higher number is better). The rules:

- You interview the candidates one at a time, **in a random order**.
- Right after each interview you must decide **on the spot**: pick this person, or let them go.
- If you let someone go, you **cannot go back** to them later.
- You know nothing about the candidates you haven't interviewed yet.
- You must end up with *somebody* - so if you reach the last person without having chosen anyone, you are stuck with that last person.

This famous puzzle is also known as the *secretary problem*, the *fussy suitor problem*, and the *best choice problem*. The best strategies all have the same shape:

> Interview the first **N** people **without choosing anyone**, and remember the highest lab skills you saw among them - call it the **benchmark**. Then keep interviewing, and choose the **first person whose lab skills beat the benchmark**.

There is a formula for the best choice of N, but we will find it **by simulation**, building up the answer one piece at a time.

**3.** Finding the highest lab skills  of the firs N students interviewed. Set `N = 10`.

- Using a `for` loop over `range(N)` and a **running maximum** (starting from `best_of_first_N = 0` and at each step in the loop update `best_of_first_N` if the current lab skills are higher than the current value of `best_of_first_N`), find the highest lab skills among the first `N` people in `interview_order`.
- Print the final vlue of `best_of_first_N`, we'll call this the benchmark lab skills.
- Every candidate in our class has lab skills above 0. Why does starting the running maximum at 0 depend on that?

```python
N = 10
best_of_first_N=0
for i in range(N):
    name=interview_order[i]
    skill=lab_partners[name]["lab_skills"]
    if skill > best_of_first_N:
        best_of_first_N=skill
print(best_of_first_N)

#Starting the running maximum at 0 works because all candidates have lab skills greater than 0. 
# This ensures that the first candidate’s skill will be greater than the initial value 
# and the running maximum will be updated. 
# If lab skills could be negative, starting at 0 could give an incorrect maximum because none of the negative values would replace 0.
```

**5.** The selection rule of the lab partner problem asks one small question over and over: *does this person's skill beat the benchmark?* Let's turn that question into a function.

- Write a function `beats_benchmark` with two parameters, `skill` and `benchmark`.
- Inside the function, write an `if`/`else` statement: if `skill > benchmark`, return `True`; otherwise return `False`.
- Using a `for` loop over the list below, apply your function to every score with `benchmark = 6.0`. At the end of the loop you should have a list of 10 elements, each either `True` or `False`.
- Bonus: `True` counts as 1 and `False` counts as 0, so `np.sum` of your list counts how many candidates beat the benchmark. Try it.

```python
import numpy as np
skill_scores = [3.24, 1.51, 0.51, 6.72, 6.72, 3.66, 0.58, 5.07, 0.37, 4.34]

# your code here

def beats_benchmark(skill, benchmark):
    if skill > benchmark:
        return True
    else:
        return False

results = []

for skill in skill_scores:
    result = beats_benchmark(skill, 6.0)
    results.append(result)

print(results)

print(np.sum(results))
```

**8.** Package the complete trial above into a function:

- Define a function `run_one_trial` with a single parameter, `N`.
- The body is the cell above: shuffle, benchmark phase, selection phase. (The function will *read* the global dictionary `lab_partners` - just like `lab_skills_of` did in Section III.)
- The function should have **two return values**: `chosen_name` and `chosen_skill`.

Which variables in your function are **local**? After restarting and running only your function cell, check that `interview_order` does not appear in your workspace until *you* create it.

```python
def run_one_trial(N):
    # your code here (adapt the cell above)
    interview_order = list(lab_partners.keys())
    random.shuffle(interview_order)

    best_of_first_N = 0

    for i in range(N):
        name = interview_order[i]
        skill = lab_partners[name]["lab_skills"]
        if skill > best_of_first_N:
            best_of_first_N = skill

    chosen_name = interview_order[-1]

    for i in range(N, len(interview_order)):
        name = interview_order[i]
        skill = lab_partners[name]["lab_skills"]
        if skill > best_of_first_N:
            chosen_name = name
            break

    chosen_skill = lab_partners[chosen_name]["lab_skills"]

    return chosen_name, chosen_skill


# test it a few times:
chosen_name, chosen_skill = run_one_trial(10)
print(f"You chose {chosen_name}, whose lab skills are {chosen_skill}.")

chosen_name, chosen_skill = run_one_trial(10)
print(f"You chose {chosen_name}, whose lab skills are {chosen_skill}.")

chosen_name, chosen_skill = run_one_trial(10)
print(f"You chose {chosen_name}, whose lab skills are {chosen_skill}.")

#interview_order, best_of_first_N, i, name, skill, chosen_name, and chosen_skill are local variables because they are created inside the function.
#  N is also local because it is a parameter of the function. 
# lab_partners is global because it is defined outside the function.
```

**10.** Now, the numerical answer. Which N gives the best average lab partner?

- Loop over every `N` in `range(num_candidates)`.
- For each N, compute the 100-trial average from Exercise 7.
- Track the winner with a **running maximum**: keep variables `best_average` and `best_N`, and update *both* whenever an average beats the current best.
- At the end, print the best N and its average.

Run the whole cell several times. Does the *same* N win every time? Next class we will draw a picture that explains what is going on.

```python
num_candidates = len(lab_partners)
num_trials = 100

best_average = 0
best_N = 0

for N in range(num_candidates):
    chosen_skills=[]
    for i in range(num_trials):
        chosen_name,chosen_skill = run_one_trial(N)
        chosen_skills.append(chosen_skill)
    average=np.mean(chosen_skills)
    if average > best_average:
        best_average=average
        best_N=N

print(best_average)
print(best_N)

        

print(f"The best N was {best_N}, with an average partner skill of {best_average}")

#No, the same N does not always win. 
# Because the interview order is randomized and we only run 100 trials for each N, 
# the average skill varies from run to run. For example, I got N = 4 in one run and N = 10 in another run.
#Increasing the number of trials would likely make the results more stable.
```
